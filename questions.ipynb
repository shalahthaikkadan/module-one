## 1D slicing

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60, 70, 80])
print(arr[2:5])
```

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60, 70, 80])
print(arr[3:])
```

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60, 70, 80])
print(arr[:4])
```

## Negative slicing

```python
import numpy as np

arr = np.array([5, 10, 15, 20, 25, 30, 35])
print(arr[-3:])
```

```python
import numpy as np

arr = np.array([5, 10, 15, 20, 25, 30, 35])
print(arr[-3:-1])
```

## Step slicing

```python
import numpy as np
arr = np.array([5, 10, 15, 20, 25, 30, 35])
print(arr[::2])
```

```python
import numpy as np
arr = np.array([5, 10, 15, 20, 25, 30, 35])
print(arr[1::3])
```
