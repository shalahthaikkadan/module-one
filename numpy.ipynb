# matrix multiplication

##### Matrix multiplication is how the network learns and transforms data.Matrix multiplication is used in AI to combine many input features with weights at once to produce outputs — this is how neural networks make predictions.This is how a neural network layer works:inputs × weights → outputs

```python
import numpy as np

A= np.array([[3,4],
            [5,4]])

B=np.array([[5,6],
            [7,6]])

print(np.matmul(A,B))
```

```python
import numpy as np

A= np.array([[3,4],
            [5,4]])

B=np.array([[5,6],
            [7,6]])

print(A@B)
```

```python
import numpy as np

A= np.array([[3,4],
            [5,4]])

B=np.array([[5,6],
            [7,6]])

print(A.dot(B))
```

## 1*3 3*2 multiplication

# Concatenation

### combing two array

### Row wise

```python
import numpy as np

A=np.array([[1,2,3]])
B=np.array([[4,5,6]])
c=np.concatenate((A,B),axis=0) #row-wise
print(c)
```

### Coloumn wise

```python
import numpy as np

A=np.array([[1,2,3]])
B=np.array([[4,5,6]])
c=np.concatenate((A,B),axis=1) #coloumn wise
print(c)
```

## Numpy filtering


```python
import numpy as np

array = np.array([[50, 60, 40, 89, 90],
                  [89, 78, 56, 75, 99]])

marks = array[array > 40]
print("All marks > 40:", marks)  
```

```python
import numpy as np

array = np.array([[50, 60, 40, 89, 90],
                  [89, 78, 56, 75, 99]])

new_marks = array[(array > 40) & (array < 90)]
print("Marks between 40 and 90:", new_marks)  
```

```python
import numpy as np

array = np.array([[50, 60, 40, 89, 90],
                  [89, 78, 56, 75, 99]])

new_marks = array[(array > 40) | (array < 90)]
print("Marks between 40 and 90:", new_marks)  
```

```python
import numpy as np

array = np.array([[50, 60, 40, 89, 90],
                  [89, 78, 56, 75, 99]])

updated_marks = array[~(array < 40)]
print(updated_marks)
```
