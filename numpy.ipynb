# NUMPY

### Numpy multiplication

```python
import numpy as np

lst = [2, 4, 6]

array = np.array(lst)
result = array * 2

print(result)


```

### Numpy addition


```python
import numpy as np

lst=[2,4,5]
lst1=[2,4,5]
array1=np.array(lst)
array2=np.array(lst1)
result=array1+array2
print(result) 
```

### Numpy Substraction


```python
import numpy as np

lst=[2,3,5]
lst1=[2,4,5]
array1=np.array(lst)
array2=np.array(lst1)
result=array1-array2
print(result)
```

### Numpy division

```python
import numpy as np


lst = [2, 4, 6]

array = np.array(lst)
result = array//2

print(result)
```

# Scalar arithmetic operation

```python
import numpy as np

#Mathematical operation in a single line

array=np.array([1,2,3,4,5])

print(array+1)
print(array-1)
print(array//1)
print(array*2)
print(array**2)
print(array**3)
```

# Vector Arithmetic operation

```python
import numpy as np


array = np.array([1,2,3,4,5])
array1=np.array([1.2,2.2,3.3,4.4,5.6])
print(np.sqrt(array))
print(np.round(array1))
print(np.floor(array1))
print(np.pi)
```

### 2D Array

```python
import numpy as np

A=np.array([[1,2],
            [2,3]])

B=np.array([[2,3],
            [7,8]])

print(A.shape)#array dimension
print(A*B)
```

# slicing 2D

## Row slicing

```python
import numpy as np

lst=np.array([[1,2,3,4,5,6,7],
              [2,4,5,6,7,8,8]])

print(lst[0,0:4])

```

## coloumn slicing
